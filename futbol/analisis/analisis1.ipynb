```python
mayor_tiempo_jugado = df.groupby("player_name")["played_time"].sum().sort_values(ascending=False).head(20)
plt.figure(figsize=(12,6))
plt.title("Jugadores con mayor tiempo en la cancha")
barras = sns.barplot(x=mayor_tiempo_jugado.index,
            y=mayor_tiempo_jugado.values,
            hue=mayor_tiempo_jugado.index,
            palette="viridis")

for contenedores in barras.containers:
    barras.bar_label(contenedores)

plt.grid(axis="y",
         linestyle="--",
         alpha=0.5)
plt.xticks(rotation=45)
plt.xlabel("Nombre del jugador")
plt.ylabel("Tiempo jugado (minutos)")
plt.show()
```
