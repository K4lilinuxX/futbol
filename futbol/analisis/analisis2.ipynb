```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Mostrar todas las columnas
pd.set_option('display.max_columns', None)
```

```python
df = pd.read_csv("laliga_jugadores_limpio.csv")
```

```python
df
```

```python
# Acortamos el nombre de Vinicius Jr
df["player_name"] = df["player_name"].str.replace("Vinicius José Paixão de Oliveira Júnior", "Vinicius Jr", regex=False)
```

```python
# Jugares con más goles totales anotados

# Agrupamos
max_goles = df.groupby("player_name")["goals"].sum().sort_values(ascending=False).head(20)

# Graficamos
plt.figure(figsize=(12,7))
plt.title("Jugadores con mayor cantidad total de goles anotados")
barras = sns.barplot(x=max_goles.index,
                     y=max_goles.values,
                     hue=max_goles.index,
                     palette="plasma")
for contenedores in barras.containers:
    barras.bar_label(contenedores)
plt.xticks(rotation=45,
           ha="right")
plt.grid(axis="y",
         linestyle="--",
         alpha=0.6)
plt.xlabel("Nombre del jugador")
plt.ylabel("Cantidad de goles anotados")
plt.show()
```

```python
# Cantidad total de goles por club

# Agrupamos
clubes_j_anotadores = df.groupby("team")["goals"].sum().sort_values(ascending=False)

# Graficamos
plt.figure(figsize=(12,7))
plt.title("Cantidad total de goles por club")
barras = sns.barplot(x=clubes_j_anotadores.index,
                     y=clubes_j_anotadores.values,
                     hue=clubes_j_anotadores.index,
                     palette="Blues")
for contenedores in barras.containers:
    barras.bar_label(contenedores)
plt.xticks(rotation=45,
           ha="right")
plt.grid(axis="y",
         linestyle="--",
         alpha=0.6)
plt.xlabel("Nombre del club")
plt.ylabel("Cantidad de goles anotados")
plt.show()
```

```python
# Data frame únicamente de delanteros (forward)
df_fd = df[df["position"] == "Forward"]

# Graficamos
plt.figure(figsize=(10,6))
plt.title("Relación entre tiempo jugado y goles anotados")
# Puntos azules
plt.scatter(df_fd["played_time"],
            df_fd["goals"])
# Linea roja
m, b = np.polyfit(df_fd["played_time"], df_fd["goals"], 1)
plt.plot(df_fd["played_time"], m * df_fd["played_time"] + b, color="red")

plt.xlabel("Minutos en la cancha")
plt.ylabel("Goles anotados")
plt.show()
```

```python
# Porteros a los que más goles les han marcado

# Data frame de porteros
df_gk = df[df["position"] == "Goalkeeper"]

# Agrupamos
top_porteros = df_gk.groupby("player_name")["goals_conceded"].sum().sort_values(ascending=False).head(20)

# Agrupamos
plt.figure(figsize=(12,6))
plt.title("Porteros con mas goles regulares recibidos")
barras = sns.barplot(x=top_porteros.index,
                     y=top_porteros.values,
                     hue=top_porteros.index,
                     palette="viridis")
plt.xticks(rotation=45,
           ha="right")
plt.xlabel("Nombre del portero")
plt.ylabel("Cantidad de goles recibidos")
plt.grid(axis="y",
         linestyle="--",
         alpha=0.6)
plt.show()
```

```python
# Relación entre tiros y goles
plt.figure(figsize=(12, 6))
plt.title("Relación entre tiros y goles de cada equipo")
plt.scatter(df["shots_total"], df["goals"], s=60)
for _, r in df.iterrows():
    plt.annotate(r["team"], (r["shots_total"], r["goals"]),
                 xytext=(5, 3), textcoords="offset points", fontsize=8)
plt.xlabel("Tiros totales")
plt.ylabel("Goles")
plt.show()
```

```python
posiciones = df["position"].value_counts()
plt.figure(figsize=(11,7))
plt.title("Cantidad de jugadores en cada posición")
barras = sns.barplot(x=posiciones.index,
                     y=posiciones.values,
                     hue=posiciones.index,
                     palette="magma")
for contenedores in barras.containers:
    barras.bar_label(contenedores)
plt.xlabel("Posición del jugador")
plt.ylabel("Cantidad de jugadores")
plt.grid(axis="y",
         linestyle="--",
         alpha=0.6)
plt.show()
```

```python
# Relación entre apariciones del jugador y goles anotados

plt.figure(figsize=(12,6))
plt.title("Relación entre apariciones del jugador y goles anotados")
sns.scatterplot(data=df, x="appearances", y="goals",
                hue="position", alpha=0.6, s=100)
plt.xlabel("Cantidad de apariciones")
plt.ylabel("Cantidad de goles")
plt.show()
```
